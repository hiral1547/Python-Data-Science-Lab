# **Descriptive Statistics Using Python**

```python
from scipy import stats
import numpy as np

# weights of the people in Gym
data = np.array([
    68, 72, 65, 70, 72, 68 ])

mean = np.mean(data)
median = np.median(data)
mode = stats.mode(data)
# Find unique values and their associated frequencies
values, counts = np.unique(data, return_counts=True)

# Identify the highest frequency and pull all corresponding values
max_count = np.max(counts)
all_modes = values[counts == max_count]
std_dev = np.std(data, ddof=1)
var = np.var(data, ddof=1)
mini = np.min(data)
maxi = np.max(data)

print("Number of values:", len(data))

print("Mean:", mean)
print("Median:", median)
print("Mode:", mode.mode)
print("All Modes:", all_modes)

print("Standard Deviation:", std_dev)
print("Variance:", var)

print("Minimum:", mini)
print("Maximum:", maxi)

print("Sum squared differences:", np.sum((data - mean) ** 2))
print("Population variance:", np.var(data))
print("Sample variance:", np.var(data, ddof=1))
```

# **Statistical Analysis of Gym Member Weights Using Python**

```python


from scipy import stats
import numpy as np

# ============================================================
# REAL-LIFE EXAMPLE: GYM MEMBER WEIGHTS
# ============================================================
# Imagine I am a gym owner.
#
# I record the weights of 6 members:
# 68, 72, 65, 70, 72, 68 kg
#
# Using NumPy and SciPy, I can quickly analyze this data
# and understand the general weight profile of my members.
# ============================================================

data = np.array([
    68, 72, 65, 70, 72, 68
])


# ============================================================
# AVERAGE WEIGHT
# ============================================================
mean = np.mean(data)

# GYM OWNER INSIGHT:
# The average member weight is about 69.17 kg.
#
# Why is this useful?
# If the gym owner is planning:
# - Nutrition programs
# - Weight-loss programs
# - Weight-training programs
# - Fitness challenges
#
# the average gives a quick idea of the typical member
# weight in this group.
# ============================================================


# ============================================================
# MEDIAN WEIGHT
# ============================================================
median = np.median(data)

# GYM OWNER INSIGHT:
# The median weight is 69 kg.
#
# The gym owner can compare the median with the average
# to understand whether a few unusually heavy or light
# members are strongly affecting the average.
# ============================================================


# ============================================================
# MOST COMMON WEIGHT
# ============================================================
mode = stats.mode(data)

# Find all values having the highest frequency
values, counts = np.unique(data, return_counts=True)

max_count = np.max(counts)
all_modes = values[counts == max_count]

# GYM OWNER INSIGHT:
# 68 kg and 72 kg are the most frequently occurring weights.
#
# This tells the gym owner which weight ranges are common
# among this group of members.
#
# This could be useful when designing:
# - Workout groups
# - Fitness challenges
# - Weight-management programs
# ============================================================


# ============================================================
# STANDARD DEVIATION
# ============================================================
std_dev = np.std(data, ddof=1)

# GYM OWNER INSIGHT:
# Standard deviation tells the owner how much the member
# weights vary around the average.
#
# A SMALL standard deviation:
# Members have relatively similar weights.
#
# A LARGE standard deviation:
# Members have a wider range of weights and may have
# very different fitness/nutrition requirements.
#
# This can help the gym owner decide whether members
# should be divided into different programs or groups.
# ============================================================


# ============================================================
# VARIANCE
# ============================================================
var = np.var(data, ddof=1)

# GYM OWNER INSIGHT:
# Variance is another measure of how spread out the
# member weights are.
#
# In practical terms, the standard deviation is usually
# easier for the gym owner to understand because it is
# expressed in kilograms.
# ============================================================


# ============================================================
# MINIMUM AND MAXIMUM WEIGHT
# ============================================================
mini = np.min(data)
maxi = np.max(data)

# GYM OWNER INSIGHT:
# Minimum weight = 65 kg
# Maximum weight = 72 kg
#
# This gives the gym owner the range of weights represented
# in this group.
#
# The owner can use this information when planning:
# - Different workout groups
# - Personalized training
# - Weight-management programs
# ============================================================


# ============================================================
# PRINT THE ANALYSIS
# ============================================================

print("Number of values:", len(data))

print("Mean:", mean)
print("Median:", median)

print("Mode:", mode.mode)
print("All Modes:", all_modes)

print("Standard Deviation:", std_dev)
print("Variance:", var)

print("Minimum:", mini)
print("Maximum:", maxi)


# ============================================================
# HOW MUCH DO THE WEIGHTS VARY FROM THE AVERAGE?
# ============================================================
print(
    "Sum squared differences:",
    np.sum((data - mean) ** 2)
)

# This is an intermediate statistical calculation used
# to calculate variance and standard deviation.
#
# A gym owner normally wouldn't look at this number directly,
# but Python uses it to calculate the spread of the data.


# ============================================================
# POPULATION VARIANCE
# ============================================================
print("Population variance:", np.var(data))

# If these 6 members are the complete group we want to study,
# this gives the variance of their weights.


# ============================================================
# SAMPLE VARIANCE
# ============================================================
print("Sample variance:", np.var(data, ddof=1))

# If these 6 members are a sample of a much larger group
# of gym members, this can be used to estimate the variation
# in the larger population.

# NumPy is a Python library used for working with arrays.
# NumPy stands for Numerical Python.
```
